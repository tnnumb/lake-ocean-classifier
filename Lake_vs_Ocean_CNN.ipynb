## Step 1 · Setup and settings

````python
!pip -q install gradio pillow-heif

import os, json, random, shutil, zipfile, glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
import tensorflow as tf
import keras
from keras import layers, regularizers
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report

try:                                   # lets you use iPhone .HEIC photos too
    import pillow_heif
    pillow_heif.register_heif_opener()
except Exception:
    pass

# ---------------- SETTINGS (write these in your report) ----------------
SEED            = 42           # fixed seed → repeatable results
IMG_SIZE        = (224, 224)   # input size for MobileNetV2
BATCH_SIZE      = 16
VAL_SPLIT       = 0.20         # 20% of each class goes to validation
DENSE_UNITS     = 128          # neurons in the hidden Dense layer
DROPOUT         = 0.4
EPOCHS_HEAD     = 30           # max epochs, stage 1 (train the new head)
LR_HEAD         = 1e-3         # Adam learning rate, stage 1
FINE_TUNE       = True         # stage 2: fine-tune the top of MobileNetV2
FINE_TUNE_LAYERS= 30           # number of top base layers to unfreeze
EPOCHS_FINE     = 15           # max epochs, stage 2
LR_FINE         = 1e-5         # Adam learning rate, stage 2 (small → stable)
PATIENCE        = 6            # early stopping patience
# ------------------------------------------------------------------------

keras.utils.set_random_seed(SEED)      # seeds Python, NumPy and TensorFlow
print("TensorFlow", tf.__version__, "| Keras", keras.__version__)
print("GPU:", tf.config.list_physical_devices('GPU') or "none (CPU is fine, just slower)")
````

## Step 2 · Load your images
Choose **one** option:
- `DATA_SOURCE = "upload"` → a file picker opens; choose your zip file.
- `DATA_SOURCE = "drive"` → set `DRIVE_PATH` to the folder that holds your images.

The code finds your folders automatically. Both layouts work:
- **Training images**: a folder with one subfolder per class (`dataset/lake`, `dataset/ocean`, or `train/lake`, `train/ocean`).
- **Test images**: a folder whose name contains `test`. It can have class subfolders (`test/lake`) **or** hold all the images
  together, with the class name in each file name (`lake_test_1.jpg`, `ocean_test_3.jpg`).

````python
DATA_SOURCE = "upload"     # "upload" or "drive"
DRIVE_PATH  = "/content/drive/MyDrive/ML_Project/lake_ocean"   # used only when DATA_SOURCE = "drive"

RAW_DIR = "/content/raw"
EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".gif", ".webp", ".tif", ".tiff", ".jfif", ".heic", ".heif"}

if DATA_SOURCE == "upload":
    from google.colab import files
    shutil.rmtree(RAW_DIR, ignore_errors=True); os.makedirs(RAW_DIR)
    uploaded = files.upload()                     # pick your zip file
    for name in uploaded:
        if name.lower().endswith(".zip"):
            with zipfile.ZipFile(name) as z:
                z.extractall(RAW_DIR)
            os.remove(name)
    SEARCH_ROOT = RAW_DIR
else:
    from google.colab import drive
    drive.mount("/content/drive")
    SEARCH_ROOT = DRIVE_PATH

def list_images(folder):
    return sorted(os.path.join(folder, f) for f in os.listdir(folder)
                  if os.path.splitext(f)[1].lower() in EXTS and not f.startswith("."))

def walk_dirs(root):
    # Every folder under root, skipping hidden folders and macOS junk
    for dirpath, dirnames, _ in os.walk(root):
        dirnames[:] = sorted(d for d in dirnames if not d.startswith((".", "__MACOSX")))
        yield dirpath, dirnames

def is_test_name(name):
    return "test" in name.lower()

def find_train_dir(root):
    # 1) a folder called train/training, 2) otherwise the first non-test folder with 2+ class subfolders of images
    for dirpath, dirnames in walk_dirs(root):
        for d in dirnames:
            if d.lower() in {"train", "training"}:
                return os.path.join(dirpath, d)
    for dirpath, dirnames in walk_dirs(root):
        if is_test_name(os.path.relpath(dirpath, root)):
            continue
        cls = [d for d in dirnames if not is_test_name(d) and list_images(os.path.join(dirpath, d))]
        if len(cls) >= 2:
            return dirpath
    return None

def find_test_dir(root):
    for dirpath, dirnames in walk_dirs(root):
        for d in dirnames:
            if is_test_name(d):
                return os.path.join(dirpath, d)
    return None

RAW_TRAIN = find_train_dir(SEARCH_ROOT)
RAW_TEST  = find_test_dir(SEARCH_ROOT)
assert RAW_TRAIN, f"Couldn't find training images inside {SEARCH_ROOT}. Need a folder with one subfolder per class (e.g. dataset/lake, dataset/ocean)."
assert RAW_TEST,  f"Couldn't find a test folder inside {SEARCH_ROOT}. Need a folder with 'test' in its name."

train_classes = {d.lower(): os.path.join(RAW_TRAIN, d) for d in sorted(os.listdir(RAW_TRAIN))
                 if os.path.isdir(os.path.join(RAW_TRAIN, d)) and not d.startswith((".", "__"))
                 and not is_test_name(d) and list_images(os.path.join(RAW_TRAIN, d))}
CLASS_NAMES = sorted(train_classes)
NUM_CLASSES = len(CLASS_NAMES)
assert NUM_CLASSES >= 2, f"Need at least 2 class folders in {RAW_TRAIN}, found: {CLASS_NAMES}"

# Test images: class subfolders (test/lake/...) OR one folder with the class name in each file name
test_items = []   # (path, class)
sub = {d.lower(): os.path.join(RAW_TEST, d) for d in os.listdir(RAW_TEST) if os.path.isdir(os.path.join(RAW_TEST, d))}
if set(CLASS_NAMES) <= set(sub):
    for c in CLASS_NAMES:
        test_items += [(p, c) for p in list_images(sub[c])]
else:
    unlabeled = []
    for p in list_images(RAW_TEST):
        fname = os.path.basename(p).lower()
        hits = [c for c in CLASS_NAMES if c in fname]
        if len(hits) == 1:
            test_items.append((p, hits[0]))
        else:
            unlabeled.append(os.path.basename(p))
    if unlabeled:
        print("⚠️ These test files were skipped (put 'lake' or 'ocean' in the file name):", unlabeled)
assert test_items, f"No labelled test images found in {RAW_TEST}"

print("Training folder:", RAW_TRAIN)
print("Test folder    :", RAW_TEST)
print("Classes        :", CLASS_NAMES)
````

## Step 3 · Clean the images and split train / validation
- **Duplicates removed:** a training photo that is an exact copy of a test photo is taken out of training.
  Otherwise the test would show photos the model has already seen, and the test score would be too high.
  Exact copies inside training are also removed.
- Every image is opened, rotated upright (phone EXIF), converted to RGB and saved as JPG. Broken files are skipped and listed.
- 20% of **each class** goes to validation (a stratified split, so both classes stay balanced).

````python
import hashlib
DATA_DIR = "/content/data"

def load_rgb(path):
    img = Image.open(path)
    img = ImageOps.exif_transpose(img)
    if img.mode in ("RGBA", "LA", "P"):
        img = img.convert("RGBA")
        bg = Image.new("RGB", img.size, (255, 255, 255))
        bg.paste(img, mask=img.split()[-1])
        return bg
    return img.convert("RGB")

def file_hash(path):
    with open(path, "rb") as f:
        return hashlib.md5(f.read()).hexdigest()

def save_clean(paths, out_dir, skipped, names=None):
    os.makedirs(out_dir, exist_ok=True)
    n = 0
    for i, p in enumerate(paths):
        try:
            img = load_rgb(p)
            img.thumbnail((800, 800))
            stem = names[i] if names else f"{n:04d}"
            img.save(os.path.join(out_dir, f"{stem}.jpg"), quality=95)
            n += 1
        except Exception as e:
            skipped.append(f"{p}  ({type(e).__name__})")
    return n

test_hashes = {file_hash(p) for p, _ in test_items}
shutil.rmtree(DATA_DIR, ignore_errors=True)
rng = random.Random(SEED)
skipped, rows, removed = [], [], []
seen = set()
for c in CLASS_NAMES:
    imgs = []
    for p in list_images(train_classes[c]):
        h = file_hash(p)
        if h in test_hashes:
            removed.append(f"{c}/{os.path.basename(p)}  (same as a test image)")
        elif h in seen:
            removed.append(f"{c}/{os.path.basename(p)}  (duplicate inside training)")
        else:
            seen.add(h); imgs.append(p)
    rng.shuffle(imgs)
    n_val = max(1, round(len(imgs) * VAL_SPLIT))
    n_va = save_clean(imgs[:n_val], f"{DATA_DIR}/val/{c}", skipped)
    n_tr = save_clean(imgs[n_val:], f"{DATA_DIR}/train/{c}", skipped)
    tps = [p for p, cc in test_items if cc == c]
    n_te = save_clean(tps, f"{DATA_DIR}/test/{c}", skipped,
                      names=[os.path.splitext(os.path.basename(p))[0] for p in tps])
    rows.append({"class": c, "train": n_tr, "validation": n_va, "test": n_te})

counts = pd.DataFrame(rows).set_index("class")
counts.loc["TOTAL"] = counts.sum()
display(counts)
print(f"Validation = {counts.loc['TOTAL','validation'] / counts.loc['TOTAL','train']:.0%} of the training images")
if removed:
    print(f"\n🧹 Removed {len(removed)} duplicate training image(s):")
    print("\n".join("  " + s for s in removed))
if skipped:
    print(f"\n⚠️ Skipped {len(skipped)} file(s) that couldn't be read:")
    print("\n".join("  " + s for s in skipped))
````

## Step 5 · Build the input pipelines

````python
def make_ds(split, shuffle):
    return keras.utils.image_dataset_from_directory(
        f"{DATA_DIR}/{split}", class_names=CLASS_NAMES, label_mode="int",
        image_size=IMG_SIZE, batch_size=BATCH_SIZE, shuffle=shuffle, seed=SEED)

AUTOTUNE = tf.data.AUTOTUNE
train_ds = make_ds("train", True).cache().shuffle(500, seed=SEED).prefetch(AUTOTUNE)
val_ds   = make_ds("val",   False).cache().prefetch(AUTOTUNE)

# Class weights keep training fair if you later add more photos to one class than the other
n_per_class = np.array([len(os.listdir(f"{DATA_DIR}/train/{c}")) for c in CLASS_NAMES])
class_weight = {i: n_per_class.sum() / (NUM_CLASSES * n) for i, n in enumerate(n_per_class)}
print("Classes:", CLASS_NAMES, "| class weights:", {CLASS_NAMES[i]: round(w, 2) for i, w in class_weight.items()})
````

## Step 6 · Build the CNN
**Why transfer learning?** With only ~100 training images, a CNN trained from zero tends to memorise the photos.
MobileNetV2 is a CNN already trained on 1.4 million ImageNet photos, so it already recognises water, sky, sand,
mountains and trees. We keep that part and train a small new classifier on top. It is fast and accurate, and gives
similar results from run to run.

| Part | Layers | Details |
|---|---|---|
| Data augmentation (training only) | 4 | random flip, rotation ±18°, zoom ±10%, shift ±5% |
| Rescaling | 1 | pixels 0–255 → −1…1 |
| MobileNetV2 base (CNN feature extractor) | 154 | ImageNet weights, frozen in stage 1 |
| Global Average Pooling | 1 | 7×7×1280 → 1280 features |
| Dropout | 1 | rate 0.4 |
| Dense (hidden) | 1 | **128 neurons**, ReLU, L2 1e-4 |
| Dropout | 1 | rate 0.4 |
| Dense (output) | 1 | **2 neurons**, softmax → percentage for lake / ocean |

````python
augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.1),
    layers.RandomTranslation(0.05, 0.05),
], name="augmentation")   # no colour changes: water colour is an important clue

base = keras.applications.MobileNetV2(input_shape=IMG_SIZE + (3,), include_top=False, weights="imagenet")
base.trainable = False

inputs  = keras.Input(shape=IMG_SIZE + (3,), name="image")
x = augmentation(inputs)
x = layers.Rescaling(1 / 127.5, offset=-1, name="rescale")(x)          # MobileNetV2 preprocessing
x = base(x, training=False)                                            # BatchNorm stays in inference mode
x = layers.GlobalAveragePooling2D(name="gap")(x)
x = layers.Dropout(DROPOUT, name="dropout_1")(x)
x = layers.Dense(DENSE_UNITS, activation="relu",
                 kernel_regularizer=regularizers.l2(1e-4), name="dense_hidden")(x)
x = layers.Dropout(DROPOUT, name="dropout_2")(x)
outputs = layers.Dense(NUM_CLASSES, activation="softmax", name="output")(x)
model = keras.Model(inputs, outputs, name="lake_ocean_cnn")

model.compile(optimizer=keras.optimizers.Adam(LR_HEAD),
              loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()
print("MobileNetV2 base layers:", len(base.layers))
````

## Step 7 · Train, stage 1 (train only the new head)
Early stopping watches the validation loss and keeps the **best** weights, so a bad last epoch can't hurt the result.

````python
def callbacks():
    return [
        keras.callbacks.EarlyStopping(monitor="val_loss", patience=PATIENCE, restore_best_weights=True, verbose=1),
        keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-7, verbose=1),
    ]

hist1 = model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS_HEAD,
                  class_weight=class_weight, callbacks=callbacks(), verbose=2)
stage1_epochs = len(hist1.history["loss"])
````

## Step 8 · Train, stage 2 (fine-tune the top of MobileNetV2)
The top layers of the base are unfrozen and trained with a **very small** learning rate, so they adapt to lakes and oceans
without forgetting what they learned. BatchNorm layers stay frozen to keep training stable.

````python
hist2 = None
if FINE_TUNE:
    base.trainable = True
    for layer in base.layers[:-FINE_TUNE_LAYERS]:
        layer.trainable = False
    for layer in base.layers:
        if isinstance(layer, layers.BatchNormalization):
            layer.trainable = False
    model.compile(optimizer=keras.optimizers.Adam(LR_FINE),
                  loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    print("Trainable weights now:", sum(np.prod(w.shape) for w in model.trainable_weights))
    hist2 = model.fit(train_ds, validation_data=val_ds, epochs=stage1_epochs + EPOCHS_FINE,
                      initial_epoch=stage1_epochs, class_weight=class_weight,
                      callbacks=callbacks(), verbose=2)
stage2_epochs = len(hist2.history["loss"]) if hist2 else 0
print(f"Epochs run: stage 1 = {stage1_epochs}, stage 2 = {stage2_epochs}, total = {stage1_epochs + stage2_epochs}")
````

## Step 9 · Training curves

````python
os.makedirs("/content/report_figures", exist_ok=True)
H = {k: hist1.history[k] + (hist2.history[k] if hist2 else []) for k in ["accuracy", "val_accuracy", "loss", "val_loss"]}
ep = np.arange(1, len(H["loss"]) + 1)

fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 4.5))
for ax, key, title in [(a1, "accuracy", "Accuracy"), (a2, "loss", "Loss")]:
    ax.plot(ep, H[key], "o-", label="Train", ms=3)
    ax.plot(ep, H["val_" + key], "s-", label="Validation", ms=3)
    if hist2: ax.axvline(stage1_epochs + 0.5, color="gray", ls="--", lw=1, label="Start fine-tuning")
    ax.set_title(title); ax.set_xlabel("Epoch"); ax.grid(alpha=0.3); ax.legend()
    ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))
a1.set_ylim(0, 1.02)
plt.tight_layout(); plt.savefig("/content/report_figures/training_curves.png", dpi=150); plt.show()
````

## Step 10 · Validation results

````python
val_loss, val_acc = model.evaluate(val_ds, verbose=0)
y_true = np.concatenate([y.numpy() for _, y in val_ds])
y_prob = model.predict(val_ds, verbose=0)
y_pred = y_prob.argmax(1)
print(f"Validation accuracy: {val_acc:.2%}   (loss {val_loss:.4f})\n")
print(classification_report(y_true, y_pred, target_names=CLASS_NAMES, digits=3, zero_division=0))

ConfusionMatrixDisplay(confusion_matrix(y_true, y_pred, labels=range(NUM_CLASSES)),
                       display_labels=CLASS_NAMES).plot(cmap="Blues", colorbar=False)
plt.title("Validation confusion matrix")
plt.savefig("/content/report_figures/confusion_validation.png", dpi=150, bbox_inches="tight"); plt.show()
````

## Step 11 · Test on the test images (percentage for each class)
Green title = correct, red = wrong.

````python
def prepare(img):
    if not isinstance(img, Image.Image):
        img = Image.fromarray(np.asarray(img).astype("uint8"))
    img = ImageOps.exif_transpose(img).convert("RGB")
    return img.resize(IMG_SIZE[::-1], Image.BILINEAR)   # same resize as training

def predict_image(img):
    # Returns {class_name: probability} for one PIL image
    arr = np.asarray(prepare(img), dtype=np.float32)[None]
    probs = model.predict(arr, verbose=0)[0]
    return {c: float(p) for c, p in zip(CLASS_NAMES, probs)}

test_rows = []
for c in CLASS_NAMES:
    for p in sorted(glob.glob(f"{DATA_DIR}/test/{c}/*.jpg")):
        probs = predict_image(Image.open(p))
        pred = max(probs, key=probs.get)
        test_rows.append({"path": p, "true": c, "pred": pred, **probs})

n = len(test_rows); ncols = 4; nrows = int(np.ceil(n / ncols))
fig, axes = plt.subplots(nrows, ncols, figsize=(4 * ncols, 5.0 * nrows), squeeze=False)
for ax in axes.flat: ax.axis("off")
for ax, r in zip(axes.flat, test_rows):
    ax.imshow(Image.open(r["path"]))
    ok = r["pred"] == r["true"]
    pct = "\n".join(f"{c}: {r[c]*100:.1f}%" for c in CLASS_NAMES)
    ax.set_title(f"True: {r['true']} | Pred: {r['pred']}\n{pct}", fontsize=10,
                 color="green" if ok else "red", fontweight="bold")
plt.tight_layout(h_pad=2.0); plt.savefig("/content/report_figures/test_predictions.png", dpi=150); plt.show()

test_df = pd.DataFrame(test_rows)
test_df["image"] = test_df["path"].map(lambda p: "/".join(p.split("/")[-2:]))
for c in CLASS_NAMES: test_df[c + " %"] = (test_df[c] * 100).round(2)
test_df["correct"] = np.where(test_df["true"] == test_df["pred"], "✅", "❌")
display(test_df[["image", "true", "pred"] + [c + " %" for c in CLASS_NAMES] + ["correct"]])
test_acc = (test_df["true"] == test_df["pred"]).mean()
print(f"\nTEST ACCURACY: {test_acc:.2%}  ({(test_df['true'] == test_df['pred']).sum()}/{n} correct)")
````

### Tips if accuracy is low
- **Add more photos.** 100+ per class makes the model clearly more stable. Use different places, weather, times of day and angles.
- **Check for confusing photos.** Lakes surrounded by mountains/forest vs open-horizon oceans with waves/beaches are the easy cases;
  a huge lake with no visible shore looks like an ocean. Keep labels honest; remove photos that even a human can't tell.
- **Avoid duplicates** between `train/` and `test/`, otherwise the test score is unrealistically high.
- Results change slightly between runs because of the GPU. The fixed seed keeps them close.
